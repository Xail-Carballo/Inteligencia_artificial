```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.utils import to_categorical
from sklearn import datasets
```

```python
iris = datasets.load_iris()
X = iris.data
target = iris.target
Y = to_categorical(target).astype("uint8")
# 0 => [1 0 0]
# 1 => [0 1 0]
# 2 => [0 0 1]
```

```python
idx = 0
print(X[idx])
print(target[idx])
print(Y[idx])
```

```python
model = keras.Sequential(
  [
    layers.Dense(3, activation="relu", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="softmax", name="layer2"),
  ]
)

model.summary()
```

```python
y = model(X)
print(y[149])
```

```python
optimizer=tf.keras.optimizers.SGD(learning_rate=0.03)
loss=tf.keras.losses.MeanSquaredError()

model.compile(optimizer, loss)
history = model.fit(X, Y, epochs=500)
```

```python
loss_final_original = history.history['loss'][-1]
print(f"--> LOSS FINAL KERAS (ORIGINAL): {loss_final_original:.4f}")
```

```python
plt.plot(history.history['loss'])
plt.show()
```

```python
pd.DataFrame(history.history).plot(figsize=(8,5))
plt.show()
```

```python
model.predict(np.array([[3,3,1,1]]))
```

```python
#nueva topolog 4x3x3x3x3
model_profundo = keras.Sequential(
  [
    layers.Dense(3, activation="relu", name="layer1", input_shape=(4,)),
    layers.Dense(3, activation="softmax", name="layer2"),
    layers.Dense(3, activation="softmax", name="layer3"),
    layers.Dense(3, activation="softmax", name="layer4"),
  ]
)

#para evidencias
print("MODELO PROFUNDO")
model_profundo.summary()

# Compilamos con los mismos paramstros
optimizer = tf.keras.optimizers.SGD(learning_rate=0.03)
loss = tf.keras.losses.MeanSquaredError()
model_profundo.compile(optimizer, loss)

# Entrenamos por 500 epochs
print("\nIniciando entrenamiento del modelo profundo...")
history_profundo = model_profundo.fit(X, Y, epochs=500, verbose=0) # verbose=0 para no imprimir 500 lineas
print("Entrenamiento completado.")

# Obtenemos el loss final y graficamos
loss_final_profundo = history_profundo.history['loss'][-1]
print(f"Loss final (Epoc 500): {loss_final_profundo:.4f}")

pd.DataFrame(history_profundo.history).plot(figsize=(8,5), title="Curva de Loss (capas extra)")
plt.show()

import time

print("\nIniciando entrenamiento del modelo profundo en Keras...")
start_time = time.time()
history_profundo = model_profundo.fit(X, Y, epochs=500, verbose=0)
end_time = time.time()
tiempo_total = end_time - start_time

print("Entrenamiento completado.")
print(f"Tiempo de entrenamiento: {tiempo_total:.2f} segundos")
```
